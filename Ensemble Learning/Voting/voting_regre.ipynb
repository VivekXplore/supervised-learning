```python
import pandas as pd 
import numpy as np
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import VotingRegressor
from sklearn.metrics import r2_score
```

```python
from sklearn.datasets import load_diabetes

data = load_diabetes()

X = data.data
y = data.target
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
lr=LinearRegression()
svm=SVR() #support vector
dt=DecisionTreeRegressor()

estimators=[('lreg',lr),('svm',svm),('tree',dt)]

# first lets do corssval 


for estimator in estimators:
  scores = cross_val_score(estimator[1],X=X_train,y=y_train,scoring='r2',cv=2)
  print(estimator[0],np.round(np.mean(scores),2))
```

```python
# very low scores now lests trying voting ensemble 


voting_ensemble_regression=VotingRegressor(estimators=estimators)
voting_ensemble_regression.fit(X_train,y_train)
ypred=voting_ensemble_regression.predict(X_test)
ypred
```

```python
r2_score(y_test,ypred)
```

```python
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```
