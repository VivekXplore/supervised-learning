````python
import pandas as pd 
import numpy as np 
from sklearn.model_selection import train_test_split,GridSearchCV,RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
````

````python
df=pd.read_csv('/Users/bibekacharya/Documents/Documents/Machine_Learning/Machine Learning/Ensemble Learning/heart.csv')
df
````

````python
X=df.drop('target',axis=1)
y=df['target']
````

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
````

````python
X_train.shape, X_test.shape
````

````python
# lets comapre how other model would compared to randomforst 
````

````python
# random forsest wihout much hyper parameteres and wihout tuning 

rf=RandomForestClassifier(random_state=42)
rf.fit(X_train,y_train)
ypred=rf.predict(X_test)
accuracy_score(y_test, ypred)
````

````python
# now lets add some hyperparamter and see performance 

rf_=RandomForestClassifier(n_estimators=5, max_features='log2', max_depth=4,criterion='entropy',bootstrap=True)
rf_.fit(X_train,y_train)
ypred_=rf_.predict(X_test)
accuracy_score(y_test,ypred_)
````

````python
# lets tune it using Gridsearch Cv first and see if accuracy increases

param={
    'n_estimators':[5,6,10,13,7],
    'max_features':['log2','sqrt',None], 
    'max_depth':[1,2,3,4,5,9],
    'criterion':['entropy','gini','log_loss'],
    'bootstrap':[True,False],
    'min_samples_split':[10,20,30,5],
    'max_leaf_nodes':[2,4,5,7,8,10,20]
}
grid=GridSearchCV(estimator=rf_,param_grid=param,cv=5,scoring='accuracy')
grid.fit(X_train,y_train)
````

Let's calculate how many combinations (models) **GridSearchCV** will train.

Your parameter grid is:

```python
param = {
    'n_estimators': [5,6,10,13,7],              # 5 values
    'max_features': ['log2','sqrt',None],       # 3 values
    'max_depth': [1,2,3,4,5,9],                 # 6 values
    'criterion': ['entropy','gini','log_loss'], # 3 values
    'bootstrap': [True,False],                  # 2 values
    'min_samples_split': [10,20,30,5],          # 4 values
    'max_leaf_nodes': [2,4,5,7,8,10,20]         # 7 values
}
```

### Step 1: Calculate Total Parameter Combinations

Multiply the number of choices for each hyperparameter:

[
5 \times 3 \times 6 \times 3 \times 2 \times 4 \times 7
]

Let's calculate:

* (5 \times 3 = 15)
* (15 \times 6 = 90)
* (90 \times 3 = 270)
* (270 \times 2 = 540)
* (540 \times 4 = 2160)
* (2160 \times 7 = 15120)

So, **15,120 different hyperparameter combinations** will be tested.

---

### Step 2: Include Cross Validation

You are using:

```python
cv = 5
```

This means **each combination is trained 5 times** (once for each fold).

Therefore,

[
15120 \times 5 = 75600
]

So the Random Forest model will be **trained 75,600 times**.

---

## Final Answer

* **Hyperparameter combinations:** **15,120**
* **Cross-validation folds:** **5**
* **Total model trainings:** **75,600**

### Conclusion

In this example, **GridSearchCV trains 75,600 models**. Since it **tries every possible combination** of the specified hyperparameters, it can be **very time-consuming**, especially on large datasets or with many parameters.

Because of this, **GridSearchCV is generally more suitable for small to medium-sized datasets**. Its advantage is that it performs an **exhaustive search**, so it often finds better hyperparameter combinations than **RandomizedSearchCV**, which tests only a subset of the possible combinations. However, the trade-off is significantly higher computation time.

### RandomizedSearchcv- 
 RandomizedSearchCV is a hyperparameter tuning technique that does not try every possible combination like GridSearchCV. Instead, it randomly selects a fixed number of combinations from the parameter space and evaluates only those.

````python

param={
    'n_estimators':[5,6,10,13,7],
    'max_features':['log2','sqrt',None], 
    'max_depth':[1,2,3,4,5,9],
    'criterion':['entropy','gini','log_loss'],
    'bootstrap':[True,False],
    'min_samples_split':[10,20,30,5],
    'max_leaf_nodes':[2,4,5,7,8,10,20]
}
random=RandomizedSearchCV(
    estimator=rf_,
    param_distributions=param,
    n_iter=20,
    cv=5
)
random.fit(X_train,y_train)
````

````python
random.best_params_
````

````python
random.best_score_
````

````python

# evaluate the tuned model on the unseen test data.

best_model = random.best_estimator_

y_pred = best_model.predict(X_test)

accuracy_score(y_test, y_pred)
````
