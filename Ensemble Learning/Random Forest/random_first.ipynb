```python
# Here we wil take a dataset and see how random forst work or what it actually means. 

import pandas as pd 
import numpy as np 
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

```

```python

X,y = make_classification(n_features=5, n_redundant=0, n_informative=5,n_clusters_per_class=1)
```

```python
df=pd.DataFrame(X,columns=['col1','col2','col3','col4','col5'])
df['target']=y
print(df.shape)
df.head()
```

This is not complete random forst but it gives idea regarding how it works .

```python
df['target'].unique()
```

```python
# we will see first classificaiton  data 
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.33, random_state=42)
```

```python
# now lets first check how single d.t will work with this data 

dt=DecisionTreeClassifier(max_depth=None) # fully grown
dt.fit(X_train,y_train)
```

```python
ypred=dt.predict(X_test)
accuracy_score(ypred,y_test)
```

```python
from sklearn.tree import plot_tree
plot_tree(dt)
```

```python
# now lets make a bagging ensemble  we will use 3 modesl 
# we are doing manually and seeing then later we will also use sklarn and see

m1=DecisionTreeClassifier()
m2=DecisionTreeClassifier()
m3=DecisionTreeClassifier()
```

the tree wil change and accuray eaxh time you do runall cuz above while creating dataset make_classification() generates a new random dataset every time you run it

```python
X_train.shape
```

```python
X_train # in numpy so we will convert it iot dataframe again so that we can do sampling 

X_train_df=pd.DataFrame(X_train)
X_test_df=pd.DataFrame(X_test)
y_train_df=pd.DataFrame(y_train)

idx1 = X_train_df.sample(67, replace=True).index
idx2 = X_train_df.sample(67, replace=True).index # idx used to keep the Notice the index connects each feature row to its correct label.
idx3 = X_train_df.sample(67, replace=True).index

X_train_df1 = X_train_df.loc[idx1]
y_train_df1 = y_train_df.loc[idx1]

X_train_df2 = X_train_df.loc[idx2]
y_train_df2 = y_train_df.loc[idx2]

X_train_df3 = X_train_df.loc[idx3]
y_train_df3 = y_train_df.loc[idx3]
```

```python
m1.fit(X_train_df1,y_train_df1)
m2.fit(X_train_df2,y_train_df2)
m3.fit(X_train_df3,y_train_df3)

m1pred=m1.predict(X_test_df)
m2pred=m2.predict(X_test_df)
m3pred=m3.predict(X_test_df)


print(accuracy_score(y_test,m1pred))
print(accuracy_score(y_test,m2pred))
print(accuracy_score(y_test,m3pred))
```
